## Funciones

### Antes de comenzar

Creamos los diccionarios como `constantes` que utilizaremos posteriormente.

```python
OPCIONES_CATALAN = {
    0: "sin conocimiento",
    1: "hasta nivel B1",
    2: "nivel B2",
    3: "C1 o superior"}

OPCIONES_EXCEL = {
    0: "nivel básico",
    1: "uso de funciones",
    2: "tablas dinámicas",
    3: "Visual Basic y macros"}

OPCIONES_TITULACION = {
    0: "ESO",
    1: "FP",
    2: "Universidad",
    3: "Máster o superior"}

ORDEN_PRIORIDAD = {
      "Prioridad alta": 3,
      "Candidato adecuado": 2,
      "No prioritario": 1}

```

### pedir_opcion

*   `pedir_opcion(pregunta_registro, opciones)`
1. Mostrar la pregunta.
2. Recorrer el diccionario de opciones y mostrar cada código con su descripción.
3. Pedir la elección con input() y devolverla sin convertir.

```python
def pedir_opcion(pregunta_registro, opciones):
  """Muestra un menú de opciones y devuelve la elección del usuario.
  pregunta_registro: str - pregunta que se muestra encima del menú
  opciones: dict - {código: descripción}
  Devuelve: str - la opción escrita por el usuario, sin convertir"""

  print(pregunta_registro)
  for codigo, descripcion in opciones.items():
    print(f"  {codigo} = {descripcion}")
  return input("Elige una opción: ")
```

### obtener_datos




*   `obtener_datos()`
1. Pedir nombre, teléfono y años de experiencia con input().
2. Pedir catalán, Excel y titulación con pedir_opcion() y su diccionario.
3. Devolver los 6 datos como texto, sin validar.

```python
def obtener_datos():
  """Pide los datos de un candidato y los devuelve sin validar.
  Devuelve: tupla de 6 str - nombre, teléfono, experiencia, catalán, Excel, titulación"""

  nombre = input("Nombre y apellidos: ")
  telefono = input("Teléfono de contacto: ")
  experiencia = input("Años de experiencia: ")
  catalan = pedir_opcion("Conocimiento de catalán: ", OPCIONES_CATALAN)
  excel = pedir_opcion("Dominio de Excel: ", OPCIONES_EXCEL)
  titulacion = pedir_opcion("Titulación académica: ", OPCIONES_TITULACION)
  return nombre, telefono, experiencia, catalan, excel, titulacion
```

### validar_datos


*  validar_datos(nombre, telefono, experiencia, catalan, excel, titulacion)

1. Limpiar nombre y teléfono con strip().
2. Si validar_nombre(nombre) es False → mensaje de error y devolver None.
3. Si validar_telefono(telefono) es False → mensaje de error y devolver None.
4. Convertir experiencia con convertir_entero().
   Si es None o es menor que 0 → mensaje de error y devolver None.
5. Convertir catalán, Excel y titulación con convertir_entero().
6. Para cada uno: si es None o validar_opcion(valor, su diccionario) es False
   → mensaje de error y devolver None.
7. Si todo es correcto → devolver los 6 datos limpios y convertidos.

```python
def validar_nombre(nombre):
  """Comprueba que el nombre no está vacío y que solo contiene letras y espacios.
  nombre: str
  Devuelve: bool"""

  return nombre.replace(" ", "").isalpha()
```

*  Funciones auxiliares de validación
 - `validar_nombre(nombre)`: devuelve True si, sin espacios, solo contiene letras.
 - `validar_telefono(telefono)`: devuelve True si tiene exactamente 9 dígitos.
 - `convertir_entero(dato)`: intenta convertir a entero; devuelve None si no es posible.
 - `validar_opcion(valor, opciones)`: devuelve True si el valor es una clave del diccionario.

```python
def validar_telefono(telefono):
  """Comprueba que el teléfono tiene 9 caracteres numéricos.
  telefono: str
  Devuelve: bool"""

  return len(telefono) == 9 and telefono.isdigit()
```

```python
def convertir_entero(dato):
  """Intenta convertir un texto a número entero.
  dato: str
  Devuelve: int si se puede convertir; None si no"""

  try:
    return int(dato)
  except ValueError:
    return None
```

```python
def validar_opcion(valor, opciones):
  """Comprueba que el valor existe como clave en el diccionario de opciones.
  valor: int
  opciones: dict - {código: descripción}
  Devuelve: bool"""

  return valor in opciones
```

```python
def validar_datos(nombre, telefono, experiencia, catalan, excel, titulacion):
  """ Limpia, convierte y valida los datos de un candidato.
  Recibe: los 6 datos tal como los devuelve obtener_datos (texto).
  Devuelve: tupla con los 6 datos limpios y convertidos si todo es válido;
  None si algún dato no es válido (y muestra un mensaje con el motivo)."""

  nombre = nombre.strip()
  telefono = telefono.strip()

  if not validar_nombre(nombre):
    print("Nombre no válido: solo puede contener letras y espacios.")
    return None

  if not validar_telefono(telefono):
    print("Teléfono no válido: debe tener exactamente 9 dígitos.")
    return None

  experiencia = convertir_entero(experiencia)
  if experiencia is None or experiencia < 0:
    print("Años de experiencia no válidos: escribe un número entero igual o mayor que 0.")
    return None

  catalan = convertir_entero(catalan)
  if catalan is None or not validar_opcion(catalan, OPCIONES_CATALAN):
    print("Conocimiento de catalán no válido: elige una opción del 0 al 3.")
    return None

  excel = convertir_entero(excel)
  if excel is None or not validar_opcion(excel, OPCIONES_EXCEL):
    print("Dominio de Excel no válido: elige una opción del 0 al 3.")
    return None

  titulacion = convertir_entero(titulacion)
  if titulacion is None or not validar_opcion(titulacion, OPCIONES_TITULACION):
    print("Titulación académica no válida: elige una opción del 0 al 3.")
    return None

  return nombre, telefono, experiencia, catalan, excel, titulacion
```

### calcular_puntuacion


*   `calcular_puntuacion(experiencia, catalan, excel, titulacion)`
1. Multiplicar los años de experiencia por 1.5.
2. Sumar los puntos de catalán, Excel y titulación.
3. Devolver la puntuación.

```python
def calcular_puntuacion(experiencia, catalan, excel, titulacion):
  """Calcula la puntuación del candidato según la fórmula experiencia*1.5 + catalan + excel + titulacion.
  Devuelve: float - la puntuación del candidato."""

  return experiencia * 1.5 + catalan + excel + titulacion
```

### clasificar_candidato


*   `clasificar_candidato(puntuacion)`
1. Si la puntuación es ≥ 12 → «Prioridad alta».
2. Si es ≥ 8 → «Candidato adecuado».
3. En otro caso → «No prioritario».

```python
def clasificar_candidato(puntuacion):
  """Clasifica al candidato según su puntuación.
  Devuelve: str - la clasificación del candidato."""

  if puntuacion >= 12:
    return "Prioridad alta"
  elif puntuacion >= 8:
    return "Candidato adecuado"
  else:
    return "No prioritario"
```

### mostrar_confirmacion_candidato

*   `mostrar_confirmacion_registro(nombre, telefono, puntuacion, prioridad)`
1. Mostrar el mensaje de registro realizado con éxito.
2. Mostrar nombre, teléfono, puntuación y prioridad.

```python
def mostrar_confirmacion_registro(nombre, telefono, puntuacion, prioridad):
  """Muestra al candidato la confirmación de su registro, su puntuación obtenida
   y su clasificación de prioridad.
   Devuelve: None"""

  print("Su registro se ha realizado con éxito.")
  print(f" Nombre: {nombre}")
  print(f" Teléfono: {telefono}")
  print(f" Puntuación: {puntuacion}")
  print(f" Prioridad: {prioridad}")
```

### ordenar_candidatos

*   `ordenar_candidatos(postulaciones)`
1. Ordenar la lista con sorted(), usando como criterio el valor de la prioridad (ORDEN_PRIORIDAD) y, si hay empate, la puntuación.
2. Ordenar de mayor a menor (reverse=True) y devolver la lista ordenada.

```python
def ordenar_candidatos(postulaciones):
  """Ordena los candidatos de mayor a menor prioridad y, en caso de empate, de mayor a menor puntuación.
  postulaciones: list de dict
  Devuelve: list de dict - la lista ordenada"""

  return sorted(postulaciones, key=lambda c: (ORDEN_PRIORIDAD[c["prioridad"]], c["puntuacion"]), reverse=True)
```

### mostrar_informe


*   `mostrar_informe(postulaciones_ordenada)`
1. Mostrar el título del informe.
2. Recorrer la lista ordenada numerando desde 1 y mostrar la posición, el nombre, el teléfono y la puntuación de cada candidato.

```python
def mostrar_informe(postulaciones_ordenada):
  """Muestra el informe final de los candidatos, ya ordenados.
  postulaciones_ordenada: list de dict
  Devuelve: None"""

  print("INFORME FINAL DE CANDIDATOS")
  for posicion, candidato in enumerate(postulaciones_ordenada, start=1):
    print(f"{posicion}. {candidato['nombre']} | Tel: {candidato['telefono']} | Puntuación: {candidato['puntuacion']}")
```

### main_seleccion

*   `main_seleccion()`
1. Crear la lista vacía postulaciones.
2. Mientras haya menos de 10 candidatos: pedir datos, validarlos y, si son válidos, calcular la puntuación, clasificar, mostrar la confirmación y guardar el candidato.
3. Al completar los 10: ordenar la lista y mostrar el informe final.

```python
def main_seleccion():
  """Orquesta el proceso de inscripción: registra candidatos hasta tener 10 válidos,
   muestra una confirmación a cada candidato registrado, y muestra el informe final
   ordenado por prioridad y puntuación.
   Devuelve: None"""

  postulaciones = []

  while len(postulaciones) < 10:
    print("\n--- Nuevo registro ---")
    nombre, telefono, experiencia, catalan, excel, titulacion = obtener_datos()
    datos_validos = validar_datos(nombre, telefono, experiencia, catalan, excel, titulacion)

    if datos_validos is not None:
      nombre, telefono, experiencia, catalan, excel, titulacion = datos_validos
      puntuacion = calcular_puntuacion(experiencia, catalan, excel, titulacion)
      prioridad = clasificar_candidato(puntuacion)

      mostrar_confirmacion_registro(nombre, telefono, puntuacion, prioridad)

      candidato = {
        "nombre": nombre,
        "telefono": telefono,
        "puntuacion": puntuacion,
        "prioridad": prioridad
      }
      postulaciones.append(candidato)

  print("\nSe han completado los 10 registros. El formulario queda cerrado.\n")
  postulaciones_ordenada = ordenar_candidatos(postulaciones)
  mostrar_informe(postulaciones_ordenada)
```
