## Funciones

### Antes de comenzar

Creamos los diccionarios como `constantes` que utilizaremos posteriormente.

```python
OPCIONES_CATALAN = {
    0: "sin conocimiento",
    1: "hasta nivel B1",
    2: "nivel B2",
    3: "C1 o superior"}

OPCIONES_EXCEL = {
    0: "nivel básico",
    1: "uso de funciones",
    2: "tablas dinámicas",
    3: "Visual Basic y macros"}

OPCIONES_TITULACION = {
    0: "ESO",
    1: "FP",
    2: "Universidad",
    3: "Máster o superior"}

ORDEN_PRIORIDAD = {
      "Prioridad alta": 3,
      "Candidato adecuado": 2,
      "No prioritario": 1}

```

### pedir_opcion

*   `pedir_opcion(pregunta_registro, opciones)`
1. Mostrar la pregunta.
2. Recorrer el diccionario de opciones y mostrar cada código con su descripción.
3. Pedir la elección con input() y devolverla sin convertir.

```python
def pedir_opcion(pregunta_registro, opciones):
  """Muestra un menú de opciones y devuelve la elección del usuario.
  pregunta_registro: str - pregunta que se muestra encima del menú
  opciones: dict - {código: descripción}
  Devuelve: str - la opción escrita por el usuario, sin convertir"""

  print(pregunta_registro)
  for codigo, descripcion in opciones.items():
    print(f"  {codigo} = {descripcion}")
  return input("Elige una opción: ")
```

### obtener_datos




*   `obtener_datos()`
1. Pedir nombre, teléfono y años de experiencia con input().
2. Pedir catalán, Excel y titulación con pedir_opcion() y su diccionario.
3. Devolver los 6 datos como texto, sin validar.

```python
def obtener_datos():
  """Pide los datos de un candidato y los devuelve sin validar.
  Devuelve: tupla de 6 str - nombre, teléfono, experiencia, catalán, Excel, titulación"""

  nombre = input("Nombre y apellidos: ")
  telefono = input("Teléfono de contacto: ")
  experiencia = input("Años de experiencia: ")
  catalan = pedir_opcion("Conocimiento de catalán: ", OPCIONES_CATALAN)
  excel = pedir_opcion("Dominio de Excel: ", OPCIONES_EXCEL)
  titulacion = pedir_opcion("Titulación académica: ", OPCIONES_TITULACION)
  return nombre, telefono, experiencia, catalan, excel, titulacion
```

### validar_datos


*  validar_datos(nombre, telefono, experiencia, catalan, excel, titulacion)

1. Limpiar nombre y teléfono con strip().
2. Si validar_nombre(nombre) es False → mensaje de error y devolver None.
3. Si validar_telefono(telefono) es False → mensaje de error y devolver None.
4. Convertir experiencia con convertir_entero().
   Si es None o es menor que 0 → mensaje de error y devolver None.
5. Convertir catalán, Excel y titulación con convertir_entero().
6. Para cada uno: si es None o validar_opcion(valor, su diccionario) es False
   → mensaje de error y devolver None.
7. Si todo es correcto → devolver los 6 datos limpios y convertidos.

```python
def validar_nombre(nombre):
  """Comprueba que el nombre no está vacío y que solo contiene letras y espacios.
  nombre: str
  Devuelve: bool"""

  return nombre.replace(" ", "").isalpha()
```

*  Funciones auxiliares de validación
 - `validar_nombre(nombre)`: devuelve True si, sin espacios, solo contiene letras.
 - `validar_telefono(telefono)`: devuelve True si tiene exactamente 9 dígitos.
 - `convertir_entero(dato)`: intenta convertir a entero; devuelve None si no es posible.
 - `validar_opcion(valor, opciones)`: devuelve True si el valor es una clave del diccionario.

```python
def validar_telefono(telefono):
  """Comprueba que el teléfono tiene 9 caracteres numéricos.
  telefono: str
  Devuelve: bool"""

  return len(telefono) == 9 and telefono.isdigit()
```

```python
def convertir_entero(dato):
  """Intenta convertir un texto a número entero.
  dato: str
  Devuelve: int si se puede convertir; None si no"""

  try:
    return int(dato)
  except ValueError:
    return None
```

```python
def validar_opcion(valor, opciones):
  """Comprueba que el valor existe como clave en el diccionario de opciones.
  valor: int
  opciones: dict - {código: descripción}
  Devuelve: bool"""

  return valor in opciones
```

```python
def validar_datos(nombre, telefono, experiencia, catalan, excel, titulacion):
  """ Limpia, convierte y valida los datos de un candidato.
  Recibe: los 6 datos tal como los devuelve obtener_datos (texto).
  Devuelve: tupla con los 6 datos limpios y convertidos si todo es válido;
  None si algún dato no es válido (y muestra un mensaje con el motivo)."""

  nombre = nombre.strip()
  telefono = telefono.strip()

  if not validar_nombre(nombre):
    print("Nombre no válido: solo puede contener letras y espacios.")
    return None

  if not validar_telefono(telefono):
    print("Teléfono no válido: debe tener exactamente 9 dígitos.")
    return None

  experiencia = convertir_entero(experiencia)
  if experiencia is None or experiencia < 0:
    print("Años de experiencia no válidos: escribe un número entero igual o mayor que 0.")
    return None

  catalan = convertir_entero(catalan)
  if catalan is None or not validar_opcion(catalan, OPCIONES_CATALAN):
    print("Conocimiento de catalán no válido: elige una opción del 0 al 3.")
    return None

  excel = convertir_entero(excel)
  if excel is None or not validar_opcion(excel, OPCIONES_EXCEL):
    print("Dominio de Excel no válido: elige una opción del 0 al 3.")
    return None

  titulacion = convertir_entero(titulacion)
  if titulacion is None or not validar_opcion(titulacion, OPCIONES_TITULACION):
    print("Titulación académica no válida: elige una opción del 0 al 3.")
    return None

  return nombre, telefono, experiencia, catalan, excel, titulacion
```

### calcular_puntuacion


*   `calcular_puntuacion(experiencia, catalan, excel, titulacion)`
1. Multiplicar los años de experiencia por 1.5.
2. Sumar los puntos de catalán, Excel y titulación.
3. Devolver la puntuación.

```python
def calcular_puntuacion(experiencia, catalan, excel, titulacion):
  """Calcula la puntuación del candidato según la fórmula experiencia*1.5 + catalan + excel + titulacion.
  Devuelve: float - la puntuación del candidato."""

  return experiencia * 1.5 + catalan + excel + titulacion
```

### clasificar_candidato


*   `clasificar_candidato(puntuacion)`
1. Si la puntuación es ≥ 12 → «Prioridad alta».
2. Si es ≥ 8 → «Candidato adecuado».
3. En otro caso → «No prioritario».

```python
def clasificar_candidato(puntuacion):
  """Clasifica al candidato según su puntuación.
  Devuelve: str - la clasificación del candidato."""

  if puntuacion >= 12:
    return "Prioridad alta"
  elif puntuacion >= 8:
    return "Candidato adecuado"
  else:
    return "No prioritario"
```

### mostrar_confirmacion_candidato

*   `mostrar_confirmacion_registro(nombre, telefono, puntuacion, prioridad)`
1. Mostrar el mensaje de registro realizado con éxito.
2. Mostrar nombre, teléfono, puntuación y prioridad.

```python
def mostrar_confirmacion_registro(nombre, telefono, puntuacion, prioridad):
  """Muestra al candidato la confirmación de su registro, su puntuación obtenida
   y su clasificación de prioridad.
   Devuelve: None"""

  print("Su registro se ha realizado con éxito.")
  print(f" Nombre: {nombre}")
  print(f" Teléfono: {telefono}")
  print(f" Puntuación: {puntuacion}")
  print(f" Prioridad: {prioridad}")
```

### ordenar_candidatos

*   `ordenar_candidatos(postulaciones)`
1. Ordenar la lista con sorted(), usando como criterio el valor de la prioridad (ORDEN_PRIORIDAD) y, si hay empate, la puntuación.
2. Ordenar de mayor a menor (reverse=True) y devolver la lista ordenada.

```python
def ordenar_candidatos(postulaciones):
  """Ordena los candidatos de mayor a menor prioridad y, en caso de empate, de mayor a menor puntuación.
  postulaciones: list de dict
  Devuelve: list de dict - la lista ordenada"""

  return sorted(postulaciones, key=lambda c: (ORDEN_PRIORIDAD[c["prioridad"]], c["puntuacion"]), reverse=True)
```

### mostrar_informe


*   `mostrar_informe(postulaciones_ordenada)`
1. Mostrar el título del informe.
2. Recorrer la lista ordenada numerando desde 1 y mostrar la posición, el nombre, el teléfono y la puntuación de cada candidato.

```python
def mostrar_informe(postulaciones_ordenada):
  """Muestra el informe final de los candidatos, ya ordenados.
  postulaciones_ordenada: list de dict
  Devuelve: None"""

  print("INFORME FINAL DE CANDIDATOS")
  for posicion, candidato in enumerate(postulaciones_ordenada, start=1):
    print(f"{posicion}. {candidato['nombre']} | Tel: {candidato['telefono']} | Puntuación: {candidato['puntuacion']}")
```

### main_seleccion

*   `main_seleccion()`
1. Crear la lista vacía postulaciones.
2. Mientras haya menos de 10 candidatos: pedir datos, validarlos y, si son válidos, calcular la puntuación, clasificar, mostrar la confirmación y guardar el candidato.
3. Al completar los 10: ordenar la lista y mostrar el informe final.

```python
def main_seleccion():
  """Orquesta el proceso de inscripción: registra candidatos hasta tener 10 válidos,
   muestra una confirmación a cada candidato registrado, y muestra el informe final
   ordenado por prioridad y puntuación.
   Devuelve: None"""

  postulaciones = []

  while len(postulaciones) < 10:
    print("\n--- Nuevo registro ---")
    nombre, telefono, experiencia, catalan, excel, titulacion = obtener_datos()
    datos_validos = validar_datos(nombre, telefono, experiencia, catalan, excel, titulacion)

    if datos_validos is not None:
      nombre, telefono, experiencia, catalan, excel, titulacion = datos_validos
      puntuacion = calcular_puntuacion(experiencia, catalan, excel, titulacion)
      prioridad = clasificar_candidato(puntuacion)

      mostrar_confirmacion_registro(nombre, telefono, puntuacion, prioridad)

      candidato = {
        "nombre": nombre,
        "telefono": telefono,
        "puntuacion": puntuacion,
        "prioridad": prioridad
      }
      postulaciones.append(candidato)

  print("\nSe han completado los 10 registros. El formulario queda cerrado.\n")
  postulaciones_ordenada = ordenar_candidatos(postulaciones)
  mostrar_informe(postulaciones_ordenada)
```

## Funciones

### Antes de comenzar

Creamos los diccionarios que utilizaremos mas adelante: `SEGEMENTOS`y `FACTORES_EDAD`.

```python
SEGMENTOS = {
  0: {"segmento": "Explorador",
      "mensaje": "¿Comparando? Te lo ponemos fácil: estos son los más vendidos de la semana.",
      "factor": 0.10},
  1: {"segmento": "Coleccionista",
      "mensaje": "Lo que guardaste ha bajado de precio. ¿Vas a dejarlo escapar?",
      "factor": 0.25},
  2: {"segmento": "Prospecto en checkout",
      "mensaje": "Estás a un clic. Hoy el envío corre de nuestra cuenta.",
      "factor": 0.50},
  3: {"segmento": "Interesado reciente",
      "mensaje": "Tu selección te espera, con devolución gratis y pago 100 % seguro.",
      "factor": 0.75},
  4: {"segmento": "Enfriado",
      "mensaje": "Te guardamos tu carrito… y un 10 % de descuento. Solo hasta el domingo.",
      "factor": 0.90}
}

FACTORES_EDAD = {"18-25": 1.20, "26-40": 1.00, "41-60": 0.85, "60+": 0.70}
```

### obtener_segmento_y_factor

* `obtener_segmento_y_factor(codigo_comportamiento)`

1. Buscar en `SEGMENTOS`, con el código de comportamiento como clave, los datos de ese código.
2. Devolver el segmento, el mensaje de retargeting y el factor de comportamiento.

```python
def obtener_segmento_y_factor(codigo_comportamiento):
  """Busca en SEGMENTOS el segmento, el mensaje de retargeting y el factor de comportamiento de un código.
  codigo_comportamiento: int
  Devuelve: str, str, float - segmento, mensaje y factor"""

  datos = SEGMENTOS[codigo_comportamiento]
  return datos["segmento"], datos["mensaje"], datos["factor"]
```

### obtener_factor_edad

*  `obtener_factor_edad(grupo_edad)`

1. Buscar en `FACTORES_EDAD` con el grupo de edad como clave.
2. Devolver el valor asociado a esa clave, es decir, el factor de edad.

```python
def obtener_factor_edad(grupo_edad):
  """Busca en FACTORES_EDAD con el grupo de edad como clave.
  grupo_edad: str
  Devuelve: float - factor de edad"""

  return FACTORES_EDAD[grupo_edad]
```

### calcular_riesgo

* `calcular_riesgo(factor_comportamiento, factor_edad)`

1. Multiplicar el factor de comportamiento por el factor de edad.
2. Redondear el resultado a 2 decimales.
3. Devolver el riesgo final, que indica el riesgo de que el cliente abandone la compra.

```python
def calcular_riesgo(factor_comportamiento, factor_edad):
  """Calcula el riesgo final multiplicando los dos factores y redondeando a 2 decimales.
  factor_comportamiento: float
  factor_edad: float
  Devuelve: float - riesgo final"""

  return round(factor_comportamiento * factor_edad, 2)
```

### procesar_clientes

* `procesar_clientes(clientes)`

1. Crear la lista vacía `clientes_procesados`.
2. Recorrer la lista `clientes` y, para cada cliente:
    - Obtener su segmento, su mensaje y su factor de comportamiento -> `obtener_segmento_y_factor()`
    - Obtener su factor de edad -> `obtener_factor_edad()`
    - Calcular su riesgo final -> `calcular_riesgo()`
    - Crear un diccionario con todos sus datos, con las claves en castellano, y añadirlo a `clientes_procesados`.
3. Devolver la lista `clientes_procesados`.

```python
def procesar_clientes(clientes):
  """Recorre los clientes, obtiene su segmento, mensaje, factores y riesgo final,
  y devuelve una lista con un diccionario por cliente procesado.
  clientes: list de dict
  Devuelve: list de dict"""

  clientes_procesados = []
  for cliente in clientes:
    segmento, mensaje, factor_comportamiento = obtener_segmento_y_factor(cliente["codi_comportament"])
    factor_edad = obtener_factor_edad(cliente["grup_edat"])
    riesgo = calcular_riesgo(factor_comportamiento, factor_edad)
    cliente_procesado = {
      "id_cliente": cliente["id_client"],
      "nombre": cliente["nom"],
      "edad": cliente["edat"],
      "grupo_edad": cliente["grup_edat"],
      "codigo_comportamiento": cliente["codi_comportament"],
      "segmento": segmento,
      "mensaje": mensaje,
      "factor_comportamiento": factor_comportamiento,
      "factor_edad": factor_edad,
      "riesgo_final": riesgo
    }
    clientes_procesados.append(cliente_procesado)
  return clientes_procesados
```

### Funciones auxiliares de consulta


1. A quién contactar primero -> ordenar la tabla por `riesgo_final` de mayor a menor con `sort_values()`.


* `contactar_primero(tabla)`: ordenar la tabla por riesgo final, de mayor a menor.

```python
def contactar_primero(tabla):
  """Ordena los clientes de mayor a menor riesgo final.
  tabla: DataFrame
  Devuelve: DataFrame ordenado"""

  return tabla.sort_values(by="riesgo_final", ascending=False)
```

2. Clientes con el mismo comportamiento y distinta edad -> ordenar por `codigo_comportamiento` y después por `riesgo_final`, para ver juntos a los clientes de cada código.

* `comparar_por_edad(tabla)`: ordenar la tabla por código de comportamiento y, dentro de cada código, por riesgo final de mayor a menor.

```python
def comparar_por_edad(tabla):
  """Ordena por código de comportamiento y, dentro de cada código, de mayor a menor riesgo final.
  tabla: DataFrame
  Devuelve: DataFrame ordenado"""

  return tabla.sort_values(by=["codigo_comportamiento", "riesgo_final"], ascending=[True, False])
```

3. Clientes con poco interés -> filtrar la tabla por el segmento "Explorador".


* `poco_interes(tabla)`: filtrar los clientes del segmento "Explorador".

```python
def poco_interes(tabla):
  """Filtra la tabla por el segmento "Explorador".
  tabla: DataFrame
  Devuelve: DataFrame filtrado"""

  return tabla[tabla["segmento"] == "Explorador"]
```

4. Cliente más rentable de recuperar -> filtrar el segmento "Prospecto en checkout", ordenar por `riesgo_final` de mayor a menor y quedarse con el primero con `.head(1)`.


* `mas_rentable(tabla)`: filtrar los clientes del segmento "Prospecto en checkout", ordenarlos por riesgo final de mayor a menor y quedarse con el primero.

```python
def mas_rentable(tabla):
  """Filtra la tabla por el segmento "Prospecto en checkout", ordena de mayor a menor riesgo final y devuelve el primero.
  tabla: DataFrame
  Devuelve: DataFrame filtrado y ordenado"""

  return tabla[tabla["segmento"] == "Prospecto en checkout"].sort_values(by="riesgo_final", ascending=False).head(1)
```

## Funciones

### Antes de comenzar

Creamos la lista `PACIENTES`y el diccionario `NIVELES_RIESGO` con los que trabajaremos mas adelante.

```python
PACIENTES = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
]


NIVELES_RIESGO = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
}
```

### normalizar_edad

* `normalizar_edad(edad)`

1. Convertir la edad de texto a número entero con `int()`.
2. Devolver la edad.

```python
def normalizar_edad(edad):
  """Convierte la edad a número.
  edad: str
  Devuelve: int"""

  return int(edad)
```

### normalizar_sintomas

* `normalizar_sintomas(sintomas)`

1. Convertir el texto de los síntomas a minúsculas y separarlos por ",".
2. Eliminar los espacios innecesarios de cada síntoma.
3. Reemplazar los espacios internos por "_", para que coincidan con las claves de `NIVELES_RIESGO`.
4. Devolver la lista de síntomas normalizados.

```python
def normalizar_sintomas(sintomas):
  """Normaliza los síntomas: minúsculas, eliminar espacios innecesarios y separar los síntomas por ","
  luego reemplaza los espacios por "_" en los nombres de los síntomas.
  sintomas: str
  Devuelve: list"""

  sintomas = sintomas.lower().split(",")
  sintomas = [sintoma.strip() for sintoma in sintomas]
  sintomas = [sintoma.replace(" ", "_") for sintoma in sintomas]
  return sintomas
```

### normalizar_datos

* `normalizar_datos(paciente)`

1. Separar el string del paciente por "|" en nombre, edad y síntomas.
2. Normalizar la edad -> `normalizar_edad()`
3. Normalizar los síntomas -> `normalizar_sintomas()`
4. Devolver el nombre, la edad y los síntomas.

```python
def normalizar_datos(paciente):
  """Normaliza los datos del paciente: nombre, edad, sintomas.
  paciente: str
  Devuelve: tuple - nombre (str), edad (int) y síntomas (list)."""

  nombre, edad, sintomas = paciente.split("|")
  edad = normalizar_edad(edad)
  sintomas = normalizar_sintomas(sintomas)
  return nombre, edad, sintomas
```

### asignar_nivel_riesgo

* `asignar_nivel_riesgo(sintomas)`

1. Crear la lista vacía `niveles_sintomas`.
2. Recorrer los síntomas y buscar el nivel de riesgo de cada uno en `NIVELES_RIESGO`.
3. Añadir cada nivel a `niveles_sintomas`.
4. Devolver la lista `niveles_sintomas`.

```python
def asignar_nivel_riesgo(sintomas):
  """Asigna el nivel de riesgo a cada síntoma del paciente, buscando en el diccionario NIVELES_RIESGO.
  sintoma: str
  Devuelve: list - nivel de riesgo de cada síntoma."""

  niveles_sintomas = []
  for sintoma in sintomas:
    niveles_sintomas.append(NIVELES_RIESGO[sintoma])
  return niveles_sintomas
```

### calcular_riesgo_total

* `calcular_riesgo_total(niveles_sintomas)`

1. Sumar los niveles de riesgo de los síntomas del paciente con `sum()`.
2. Devolver el riesgo total.

```python
def calcular_riesgo_total(niveles_sintomas):
  """Suma los niveles de riesgo de los síntomas de un paciente.
  niveles_sintomas: list
  Devuelve: int - riesgo total"""

  riesgo_total = sum(niveles_sintomas)
  return riesgo_total
```

### convertir_lista_df

* `convertir_lista_df(listado_general_pacientes)`

1. Convertir la lista de pacientes en una tabla con `pd.DataFrame()`.
2. Devolver la tabla.

```python
def convertir_lista_df(listado_general_pacientes):
  """Convierte la lista de pacientes en una tabla con pd.DataFrame().
  listado_general_pacientes: list
  Devuelve: DataFrame"""

  return pd.DataFrame(listado_general_pacientes)
```

### ordenar_pacientes

* `ordenar_pacientes(tabla)`

1. Ordenar la tabla por `riesgo_total` de mayor a menor con `sort_values()`.
2. Numerar el índice desde 1 según el orden de prioridad.
3. Devolver la tabla ordenada.

```python
def ordenar_pacientes(tabla):
  """Ordena la tabla de pacientes por riesgo total de mayor a menor y numerar por indice desde el 1.
  tabla: DataFrame
  Devuelve: DataFrame ordenado"""

  tabla_ordenada = tabla.sort_values("riesgo_total", ascending=False)
  tabla_ordenada = tabla_ordenada.reset_index(drop=True)
  tabla_ordenada.index = tabla_ordenada.index + 1
  return tabla_ordenada
```

### main_pacientes

* `main_pacientes(pacientes)`

1. Crear la lista vacía `listado_general_pacientes`.
2. Para cada paciente:
    - Separar y normalizar sus datos -> `normalizar_datos()`
    - Asignar a cada síntoma su nivel de riesgo -> `asignar_nivel_riesgo()`
    - Calcular su riesgo total -> `calcular_riesgo_total()`
    - Crear un diccionario con nombre, edad, síntomas y riesgo total, y añadirlo a `listado_general_pacientes`.
3. Convertir la lista en una tabla -> `convertir_lista_df()` y mostrarla.
4. Ordenar a los pacientes por riesgo total -> `ordenar_pacientes()` y mostrar el orden de prioridad de atención.

```python
def main_pacientes(pacientes):
  """Orquesta el proceso de pacientes: normaliza los datos de los pacientes, calcula el riesgo total
  de cada uno y muestra la tabla final con los datos de todos los pacientes, y luego la muestra ordenada
  por riesgo total de mayor a menor.
  pacientes: list
  Devuelve: nada (solo muestra)"""

  listado_general_pacientes = []
  for paciente in pacientes:
    nombre, edad, sintomas = normalizar_datos(paciente)
    niveles_sintomas = asignar_nivel_riesgo(sintomas)
    riesgo_total = calcular_riesgo_total(niveles_sintomas)
    paciente_normalizado = {"nombre": nombre, "edad": edad, "sintomas": sintomas, "riesgo_total": riesgo_total}
    listado_general_pacientes.append(paciente_normalizado)
  tabla = convertir_lista_df(listado_general_pacientes)
  print("Tabla de pacientes normalizados")
  display(tabla)
  print()
  tabla_ordenada = ordenar_pacientes(tabla)
  print("Tabla de pacientes ordenados por riesgo")
  display(tabla_ordenada)
```
